```python
def tambah(a, b):
  return a+b

def tolak(a, b):
  return a-b

print("Tambah = ", tambah(10, 5))
print("Tolak = ", tolak(10, 5))
```

```python
def kira_luas(panjang,lebar):
  return panjang*lebar

luas =kira_luas(5,5)
luas2 =kira_luas(4,5)
print(f"Luas = {luas}")
print(f"Luas ={luas2}")
```

```python
def status(markah):

  if markah >= 50:
    print("Anda lulus")
  else:
    print("Anda gagal")

status(75)
status(45)
```

```python
def kira_bmi(kg,m):
  return kg/(m*m)

bmi =kira_bmi(50,1.70)
print(f"BMI anda ialah: {bmi}")
```
