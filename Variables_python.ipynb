```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Nama':["Azlan","Awang","Azzam"],
    'Markah':[67,67,67]
}

df =pd.DataFrame(data) #buat data frame

#untuk visual bar
plt.bar(df['Nama'],df['Markah'],color='blue')
plt.title('Markah Pelajar')
plt.xlabel('Nama')
plt.ylabel('Markah')
plt.show

print(df)
print("Mean markah: ", df['Markah'].mean())
print("Max markah: ", df['Markah'].max())
print("Sum markah: ", df['Markah'].sum())
```
